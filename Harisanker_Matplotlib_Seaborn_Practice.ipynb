# 🟠 Part A: Matplotlib Questions

### Question 1 — Line Plot  `Matplotlib`

Calculate the **average marks** of all students in each of the four tests (`Test_1` to `Test_4`). Create a **line plot** to show how the class average changes from test to test. Add a title and axis labels.

```python
# ✏️ Your code here
test_avg = df[["Test_1", "Test_2", "Test_3","Test_4"]].mean()

plt.plot(test_avg.index, test_avg.values, marker="o")
plt.title("Test Average Trend with Tests")
plt.xlabel("Tests")
plt.ylabel("Marks")
plt.show()
```

### Question 2 — Bar Chart  `Matplotlib`

Create a **bar chart** comparing the final `Marks` of the **first five students** in the dataset.

```python
# ✏️ Your code here
first_five = df["Student"].head(5)
first_five_marks = df["Marks"].head(5)

plt.bar(first_five, first_five_marks)
plt.title("Marks of First Five Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

### Question 3 — Histogram  `Matplotlib`

Create a **histogram** of the final `Marks` of all 30 students. Change the **number of bins** (for example 3, 5 and 10) and write what difference you observe.

```python
# ✏️ Your code here
plt.hist(df["Marks"], bins=10, edgecolor="black")
plt.title("Histogram of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

**🔎 Observation:** *(when given small values for bins, we get wider bars and viceversa, more bins felt easier to understand the plot)*

### Question 4 — Scatter Plot  `Matplotlib`

Create a **scatter plot** showing the relationship between `Study_Hours` and `Marks`. Does more study time lead to higher marks?

```python
# ✏️ Your code here
plt.scatter(df["Study_Hours"], df["Marks"], color="tomato")
plt.title("Relationship between Study Hours and Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()

# Observation - More study time leads to higher marks
```

### Question 5 — Pie Chart  `Matplotlib`

Create a **pie chart** showing the **percentage of students** who prefer each programming language (`Preferred_Language`). Display the percentage on each slice.

```python
# ✏️ Your code here
language = df["Preferred_Language"].value_counts()
plt.pie(language, labels=language.index, autopct="%1.2f%%")
plt.show()
```

# 🟢 Part B: Seaborn Questions

### Question 6 — Count Plot  `Seaborn`

Using **Seaborn**, create a **count plot** showing the number of students in each `Class`.

```python
# ✏️ Your code here
sns.countplot(data=df, x="Class", color="tomato")
plt.title("Number of students in each class")
plt.ylabel("Number of Students")
plt.show()
```

### Question 7 — Box Plot  `Seaborn`

Using **Seaborn**, create a **box plot** to compare the final `Marks` of **Class A** and **Class B**. Which class has the higher median?

```python
# ✏️ Your code here
sns.boxplot(data=df, x="Class", y="Marks", color="brown")
plt.title("Marks of Class A and B")
plt.show()
```

### Question 8 — Heatmap  `Seaborn`

Create a **heatmap** of the correlation between `Study_Hours`, `Attendance`, `Sleep_Hours` and `Marks`. Show the values on the heatmap. Which column has the strongest relationship with `Marks`?

```python
# ✏️ Your code here
numeric_df = df[["Study_Hours", "Attendance", "Sleep_Hours", "Marks"]]

correlation = numeric_df.corr()

sns.heatmap(correlation, annot=True, cmap="PuOr")
plt.title("Correlation Heat Map")
plt.show()
```

**Plot 1 — Bar plot: average marks by preferred language**

**Plot 2 — Scatter plot: sleep hours vs marks**

**🔎 Observation 2:** *Here perfomance varies showing sleep does not necessarily depend the marks scored*

**Plot 3 — Line plot: test averages for each class**

```python
# ✏️ Your code here
averages = df.groupby("Class")[["Test_1", "Test_2", "Test_3","Test_4"]].mean()

plt.plot(averages.columns, averages.loc["A"], marker="s", label="Class A")
plt.plot(averages.columns, averages.loc["B"], marker="s", label="Class B")

plt.title("Test Average Comaprison")
plt.xlabel("Tests")
plt.ylabel("Average MArks")
plt.legend()
plt.grid(True)
plt.show()
```

**🔎 Observation 3:** *Class A scores better than Class B. Both classes have highest averages in test 4*
